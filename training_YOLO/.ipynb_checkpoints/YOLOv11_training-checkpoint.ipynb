# Training YOLO models

## Training with one dataset

To train the model with one dataset, use the data.yaml file that is provided with each dataset. It is recommended to provide the full path in the data.yaml file.

```python
from ultralytics import YOLO

model = YOLO('yolo11m.pt')  # If you want to start training from the begginning
#model = YOLO('YOLO11m_e15_img1920_oneClass.pt') # Continue training the achived model

# Train the model
model.train(data='data.yaml', 
            name="yolo11m_training_one",
            imgsz=1920,
            epochs=15, 
            batch=4,
            max_det=10,
            flipud=0.0,
            fliplr=0.0,
            patience=10,
           )
```

## Training with multiple datasets

To train the model with multiple datasets, use the data_multiple_datasets.yaml file. It is recommended to provide the full paths to the file.

The image YOLO_models_comparisson.png shows different trainings of YOLO model.

```python

model = YOLO('yolo11m.pt')  # If you want to start training from the begginning
#model = YOLO('YOLO11m_e15_img1920_oneClass.pt') # Continue training the achived model

# Train the model
model.train(data='data_multiple_datasets.yaml', 
            name="yolo11m_training_multiple",
            imgsz=1920,
            epochs=15, 
            batch=4,
            max_det=10,
            flipud=0.0,
            fliplr=0.0,
            patience=10,
           )
```
